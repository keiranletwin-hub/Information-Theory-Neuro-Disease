# Dataset 1 — Information-Theoretic Characterization and Prediction of Epileptic State Transitions

**Project:** Information-Theoretic Sensor Design for Pathological State Characterization and Transition Control  
**Primary Dataset:** CHB-MIT Scalp EEG Database  
**Source:** https://physionet.org/content/chbmit/1.0.0/  
**Experimental Model:** Epileptic seizure dynamics  
**Proposed Translational Application:** Glioblastoma-associated electrophysiological dynamics  
**Status:** Dataset evaluation

## 1. Research Objective

To investigate whether pathological physiological states and their transitions can be characterized and predicted using information-theoretic analysis of electrophysiological measurements.

The study will evaluate whether conditional information analysis can identify a minimal configuration of physical EEG sensors that preserves sufficient information to:

1. Distinguish interictal and ictal physiological conditions.
2. Detect informative changes in electrophysiological activity preceding seizure onset.
3. Estimate the probability of entering a pathological state within a defined prediction horizon.
4. Preserve state-relevant and transition-predictive information using a reduced electrode configuration.

A subsequent computational simulation will investigate whether appropriately timed, minimal perturbations can alter modeled transition probabilities and reduce pathological state occupancy.

The longer-term research direction is to investigate whether this framework can be adapted to electrophysiological recordings from glioblastoma and its surrounding neural microenvironment.

## 2. Central Hypothesis

Pathological physiological conditions may exhibit distinguishable probability distributions over measurable electrophysiological variables.

Furthermore, the temporal evolution of these measurements may contain information about impending transitions into pathological states.

Although electrophysiological recordings are high-dimensional, the information required for state characterization and transition prediction may be concentrated within a smaller subset of measurements.

Conditional information evaluation may therefore identify reduced physical sensing configurations that preserve clinically relevant information about both present and future physiological conditions.

## 3. Mathematical Representation of Physiological States

Let the measured physiological state vector be:

$$
X(t)=[X_1(t),X_2(t),\ldots,X_n(t)]
$$

where each component represents a measurable electrophysiological variable, such as EEG spectral power, frequency ratios, spatial relationships, or signal complexity.

A physiological condition is represented statistically by:

$$
p(X\mid D)
$$

where $D$ denotes the physiological condition.

Examples include:

$$
p(X\mid D=\text{interictal})
$$

$$
p(X\mid D=\text{preictal})
$$

$$
p(X\mid D=\text{ictal})
$$

The preictal condition will be operationally defined using a specified interval preceding annotated seizure onset. Its existence as a statistically distinguishable physiological state will be evaluated rather than assumed.

For dynamical analysis:

$$
p(X_{t+\Delta t}\mid X_t)
$$

describes the probability of future physiological measurements conditional on the present measurements.

For seizure prediction, define a future event indicator:

$$
Y_{t,\tau}=
\begin{cases}
1 & \text{if seizure onset occurs within }(t,t+\tau]\\
0 & \text{otherwise}
\end{cases}
$$

We will investigate:

$$
p(Y_{t,\tau}=1\mid X_t)
$$

where $\tau$ is the prediction horizon.

This distinguishes prediction of a clinically defined event from prediction of ordinary fluctuations in EEG activity.

## 4. Research Questions

1. Do interictal and ictal EEG recordings exhibit statistically distinguishable physiological probability distributions?
2. Does electrophysiological activity preceding seizure onset exhibit measurable differences from interictal activity?
3. How much information about the present physiological condition is contained in individual EEG measurements?
4. How much additional information does each measurement contribute after conditioning on measurements already selected?
5. What is the minimum physical electrode configuration required to preserve a specified level of state-relevant information?
6. Can a reduced electrode configuration preserve information predictive of seizure onset within a defined future interval?
7. In a simulated dynamical system, can appropriately timed minimal perturbations reduce the probability of entering a pathological state region?

## 5. Information-Theoretic Methods

### 5.1 Shannon Entropy

$$
H(X)=-\sum_x p(x)\log_2p(x)
$$

**Purpose:** Quantify uncertainty in discretized electrophysiological measurements.

### 5.2 Joint Entropy

$$
H(X,Y)=-\sum_{x,y}p(x,y)\log_2p(x,y)
$$

**Purpose:** Characterize uncertainty associated with multiple measurements considered together.

### 5.3 Conditional Entropy

$$
H(D\mid X)=H(D,X)-H(X)
$$

**Purpose:** Quantify remaining uncertainty about the physiological condition after observing a measurement.

### 5.4 Mutual Information

$$
I(X;D)=H(D)-H(D\mid X)
$$

**Purpose:** Quantify how much information a measurement provides about the physiological condition.

For transition prediction, we will also evaluate information about the future seizure-onset indicator.

### 5.5 Conditional Mutual Information

$$
I(X_i;D\mid X_S)
$$

where $X_i$ represents a candidate measurement and $X_S$ represents measurements already selected.

**Purpose:** Determine the additional information contributed by a candidate electrode after accounting for the existing sensing configuration.

For prediction, the condition variable may be replaced with the future seizure-onset indicator $Y_{t,\tau}$.

### 5.6 Kullback–Leibler Divergence

$$
D_{\mathrm{KL}}(P\|Q)
=\sum_xP(x)\log_2\frac{P(x)}{Q(x)}
$$

**Purpose:** Quantify directional statistical divergence between electrophysiological probability distributions.

KL divergence is asymmetric and is not a distance metric.

### 5.7 Jensen–Shannon Divergence

$$
D_{\mathrm{JS}}(P,Q)
=\frac12D_{\mathrm{KL}}(P\|M)
+\frac12D_{\mathrm{KL}}(Q\|M)
$$

where:

$$
M=\frac12(P+Q)
$$

**Purpose:** Provide a symmetric, bounded measure of distributional distinguishability.

### 5.8 Information Bottleneck

$$
\min_{p(z\mid x)}
I(X;Z)-\beta I(Z;D)
$$

**Purpose:** Investigate compressed representations that preserve information relevant to physiological state or future seizure onset.

### 5.9 Entropy Rate

$$
h_X=\lim_{T\to\infty}
\frac{H(X_1,\ldots,X_T)}{T}
$$

**Purpose:** Characterize uncertainty generated by the electrophysiological process over time.

Entropy rate will be investigated as a descriptor of temporal dynamics, not assumed to be a direct measure of seizure risk.

### 5.10 Transfer Entropy

$$
T_{X\to Y}
=I(Y_{t+1};X_t^{(k)}\mid Y_t^{(l)})
$$

**Purpose:** Evaluate whether the history of one electrophysiological variable provides additional predictive information about another variable's future.

Transfer entropy measures directed predictive dependence, not necessarily causation.

### 5.11 Rate–Distortion Theory

$$
R(D)=\min_{p(\hat{x}\mid x):\,\mathbb{E}[d(X,\hat X)]\leq D}
I(X;\hat X)
$$

**Purpose:** Explore tradeoffs between sensing or representation complexity and allowable loss of physiologically relevant information.

## 6. Information-Theoretic Sensor Optimization

Let $S$ represent a selected set of physical EEG electrodes and $X_S$ their acquired measurements.

For physiological state characterization:

$$
\min_S |S|
$$

subject to:

$$
I(X_S;D)\geq I_{\mathrm{target}}
$$

For transition prediction, an analogous objective is:

$$
\min_S |S|
$$

subject to:

$$
I(X_S;Y_{t,\tau})\geq I_{\mathrm{target,pred}}
$$

These formulations describe the intended optimization objectives. Information thresholds and their statistical estimability must be established during analysis.

We will also evaluate electrode placement, signal quality, measurement redundancy, and predictive performance.

The number of extracted EEG features is not equivalent to the number of physical electrodes required to acquire them.

## 7. Dataset Description

The CHB-MIT Scalp EEG Database contains long-duration EEG recordings from pediatric patients with epilepsy, including recordings with annotated seizure onset and termination times.

The dataset offers the opportunity to investigate physiological activity before, during, and after documented seizures.

**Dataset characteristics requiring inspection:**

- Available patients and recordings
- Recording duration and continuity
- Seizure onset and termination annotations
- EEG electrode identities and channel consistency
- Sampling frequency and signal quality
- Number of seizures per patient
- Availability of sufficient pre-seizure recording intervals
- Potential confounding effects and temporal data leakage

## 8. Experimental Workflow

**Experiment 1 — Physiological State Representation**

Extract EEG features and estimate condition-dependent probability distributions.

**Experiment 2 — Distributional Distinguishability**

Compare interictal and ictal distributions, and evaluate whether operationally defined preictal intervals exhibit distinguishable statistical characteristics.

**Experiment 3 — Information Content**

Estimate entropy, mutual information, and conditional mutual information for candidate electrode measurements.

**Experiment 4 — Sensor Optimization**

Select increasingly compact physical electrode configurations and evaluate retained information and predictive performance.

**Experiment 5 — Temporal Prediction**

Investigate whether EEG measurements contain information about seizure onset within defined prediction horizons.

Evaluation will use temporally appropriate separation of training and testing data, with patient-level validation where feasible.

**Experiment 6 — Simulated Intervention**

Construct a stochastic dynamical model and investigate whether state-dependent perturbations modify transition probabilities and reduce pathological state occupancy.

Simulated intervention results will be distinguished from empirical EEG findings.

## 9. Scientific Limitations

Seizure onset provides an observable pathological event, but the existence of a reliably identifiable preictal state is not assumed.

Repeated EEG windows from the same patient are statistically dependent and cannot be treated as independent patients.

Recording artifacts, patient identity, electrode differences, and temporal proximity to seizures may confound apparent prediction.

Information-theoretic association does not establish causation or therapeutic controllability.

A simulated reduction in pathological state occupancy does not establish that an equivalent intervention would prevent seizures in patients.

Epilepsy and glioblastoma have different biological mechanisms and temporal dynamics. Transfer of the sensing and prediction methodology to glioblastoma would require separate experimental validation.

## 10. Intended Proof-of-Concept Outputs

1. Probability distributions characterizing interictal and ictal electrophysiological conditions.
2. Quantitative comparison of distributional distinguishability.
3. Assessment of information preceding annotated seizure onset.
4. Conditional-information-based physical electrode selection.
5. Information-retention and predictive-performance curves as electrodes are removed.
6. A proposed minimal EEG sensing configuration.
7. A separate computational simulation of pathological state transitions and timed perturbations.

**Overall objective:** Establish whether information-theoretic analysis of real electrophysiological recordings can provide a principled foundation for compact physiological sensor design, pathological transition prediction, and future state-dependent intervention strategies, with glioblastoma-associated electrophysiology as a proposed translational research direction.

```python
import re
import pandas as pd

summary = response.text

pattern = (
    r"File Name: (.+?\.edf).*?"
    r"Number of Seizures in File: (\d+)"
    r"(.*?)(?=File Name:|\Z)"
)

records = []

for filename, count, details in re.findall(
    pattern, summary, flags=re.DOTALL
):
    count = int(count)

    starts = re.findall(
        r"Seizure(?: \d+)? Start Time: (\d+) seconds",
        details
    )

    ends = re.findall(
        r"Seizure(?: \d+)? End Time: (\d+) seconds",
        details
    )

    for start, end in zip(starts, ends):
        records.append({
            "file": filename,
            "onset_seconds": int(start),
            "end_seconds": int(end),
            "duration_seconds": int(end) - int(start)
        })

seizures_chb01 = pd.DataFrame(records)

display(seizures_chb01)

print("Total annotated seizures:", len(seizures_chb01))
```

```python

import requests
import re
import pandas as pd

base_url = "https://physionet.org/files/chbmit/1.0.0"

patient_records = []
event_records = []
failed_cases = []

for n in range(1, 25):
    patient = f"chb{n:02d}"
    url = f"{base_url}/{patient}/{patient}-summary.txt"

    try:
        response = requests.get(url, timeout=30)
        response.raise_for_status()
    except requests.RequestException as e:
        failed_cases.append((patient, str(e)))
        continue

    summary = response.text

    blocks = re.split(r"(?=File Name:)", summary)

    file_count = 0
    seizure_count = 0
    durations = []

    for block in blocks:
        name_match = re.search(r"File Name:\s*(\S+\.edf)", block)

        if not name_match:
            continue

        file_count += 1
        filename = name_match.group(1)

        count_match = re.search(
            r"Number of Seizures in File:\s*(\d+)", block
        )
        reported_count = int(count_match.group(1)) if count_match else 0

        starts = re.findall(
            r"Seizure(?: \d+)? Start Time:\s*(\d+) seconds",
            block
        )
        ends = re.findall(
            r"Seizure(?: \d+)? End Time:\s*(\d+) seconds",
            block
        )

        if len(starts) != reported_count or len(ends) != reported_count:
            print(f"Annotation mismatch: {patient} / {filename}")

        for start, end in zip(starts, ends):
            start, end = int(start), int(end)
            duration = end - start

            event_records.append({
                "patient": patient,
                "file": filename,
                "onset_seconds": start,
                "end_seconds": end,
                "duration_seconds": duration,
                "pre_seizure_minutes_in_file": round(start / 60, 2)
            })

            seizure_count += 1
            durations.append(duration)

    patient_records.append({
        "patient": patient,
        "recordings": file_count,
        "seizures": seizure_count,
        "mean_seizure_duration_s": (
            round(sum(durations) / len(durations), 1)
            if durations else None
        ),
        "min_seizure_duration_s": min(durations) if durations else None,
        "max_seizure_duration_s": max(durations) if durations else None
    })

patients_summary = pd.DataFrame(patient_records)
all_seizures = pd.DataFrame(event_records)

display(patients_summary)

print("\nTotal cases:", len(patients_summary))
print("Total recordings listed:", patients_summary["recordings"].sum())
print("Total seizures:", len(all_seizures))
print("Cases with seizures:", (patients_summary["seizures"] > 0).sum())
print("Cases with at least 3 seizures:",
      (patients_summary["seizures"] >= 3).sum())
print("Failed cases:", failed_cases)

print("\nEVENT-LEVEL TABLE")
display(all_seizures)
```

```python
import numpy as np
import pandas as pd
import re
import requests

# ============================================================
# CHB-MIT: COMPREHENSIVE PATIENT-LEVEL METADATA TABLE
# ============================================================

base_url = "https://physionet.org/files/chbmit/1.0.0"

# ------------------------------------------------------------
# 1. RETRIEVE RECORDING METADATA
# ------------------------------------------------------------

recording_rows = []

for patient in cohort["patient"]:

    url = f"{base_url}/{patient}/{patient}-summary.txt"

    response = requests.get(url, timeout=30)
    response.raise_for_status()

    blocks = re.split(r"(?=File Name:)", response.text)

    for block in blocks:

        file_match = re.search(
            r"File Name:\s*(\S+\.edf)",
            block
        )

        if not file_match:
            continue

        filename = file_match.group(1)

        start_match = re.search(
            r"File Start Time:\s*(\d+):(\d+):(\d+)",
            block
        )

        end_match = re.search(
            r"File End Time:\s*(\d+):(\d+):(\d+)",
            block
        )

        duration_hours = np.nan

        if start_match and end_match:

            h1, m1, s1 = map(int, start_match.groups())
            h2, m2, s2 = map(int, end_match.groups())

            start_seconds = h1 * 3600 + m1 * 60 + s1
            end_seconds = h2 * 3600 + m2 * 60 + s2

            duration_seconds = (
                end_seconds - start_seconds
            ) % 86400

            duration_hours = duration_seconds / 3600

        recording_rows.append({
            "patient": patient,
            "file": filename,
            "duration_hours": duration_hours
        })

recordings_df = pd.DataFrame(recording_rows)

# Remove duplicate recording entries
recordings_df = recordings_df.drop_duplicates(
    subset=["patient", "file"]
)

# ------------------------------------------------------------
# 2. CALCULATE RECORDING EXPOSURE
# ------------------------------------------------------------

exposure = (
    recordings_df
    .groupby("patient")
    .agg(
        recordings_verified=("file", "nunique"),
        recorded_hours=("duration_hours", "sum"),
        recordings_missing_duration=(
            "duration_hours",
            lambda x: x.isna().sum()
        )
    )
    .reset_index()
)

# ------------------------------------------------------------
# 3. CALCULATE SEIZURE CHARACTERISTICS
# ------------------------------------------------------------

seizure_stats = (
    all_seizures
    .groupby("patient")
    .agg(
        seizures_verified=("duration_seconds", "count"),

        mean_duration_s=("duration_seconds", "mean"),
        median_duration_s=("duration_seconds", "median"),

        min_duration_s=("duration_seconds", "min"),
        max_duration_s=("duration_seconds", "max"),

        sd_duration_s=("duration_seconds", "std"),

        mean_preseizure_min=(
            "pre_seizure_minutes_in_file", "mean"
        ),

        min_preseizure_min=(
            "pre_seizure_minutes_in_file", "min"
        ),

        files_with_seizures=("file", "nunique")
    )
    .reset_index()
)

# ------------------------------------------------------------
# 4. CALCULATE SEIZURE CLUSTERING BY FILE
# ------------------------------------------------------------

seizures_per_file = (
    all_seizures
    .groupby(["patient", "file"])
    .size()
    .groupby("patient")
    .max()
    .rename("max_seizures_per_file")
    .reset_index()
)

# ------------------------------------------------------------
# 5. MERGE ALL PATIENT VARIABLES
# ------------------------------------------------------------

master = (
    cohort[
        [
            "patient",
            "subject_id",
            "age_years",
            "sex"
        ]
    ]
    .merge(exposure, on="patient", how="left")
    .merge(seizure_stats, on="patient", how="left")
    .merge(seizures_per_file, on="patient", how="left")
)

# ------------------------------------------------------------
# 6. DERIVED VARIABLES
# ------------------------------------------------------------

# Seizure incidence normalized by recorded exposure
master["seizures_per_hour"] = (
    master["seizures_verified"]
    / master["recorded_hours"].replace(0, np.nan)
)

# Relative variability in seizure duration
master["duration_cv"] = (
    master["sd_duration_s"]
    / master["mean_duration_s"].replace(0, np.nan)
)

# Percentage of recording files containing seizures
master["percent_files_with_seizures"] = (
    100
    * master["files_with_seizures"]
    / master["recordings_verified"]
)

# ------------------------------------------------------------
# 7. CLINICAL VARIABLES NOT AVAILABLE
# ------------------------------------------------------------

# Preserve missing clinical information explicitly.
# Do not infer medication or severity from EEG.

master["medication"] = pd.NA
master["medication_dose"] = pd.NA
master["clinical_severity_score"] = pd.NA
master["seizure_type"] = pd.NA

# ------------------------------------------------------------
# 8. ORGANIZE COLUMN ORDER
# ------------------------------------------------------------

columns = [
    "patient",
    "subject_id",
    "age_years",
    "sex",

    "recordings_verified",
    "recorded_hours",
    "recordings_missing_duration",

    "seizures_verified",
    "seizures_per_hour",
    "percent_files_with_seizures",

    "mean_duration_s",
    "median_duration_s",
    "min_duration_s",
    "max_duration_s",
    "sd_duration_s",
    "duration_cv",

    "mean_preseizure_min",
    "min_preseizure_min",

    "files_with_seizures",
    "max_seizures_per_file",

    "medication",
    "medication_dose",
    "clinical_severity_score",
    "seizure_type"
]

master = master[columns].round(3)

# ------------------------------------------------------------
# 9. DISPLAY MASTER TABLE
# ------------------------------------------------------------

print("CHB-MIT MASTER PATIENT TABLE")
display(master)

# ------------------------------------------------------------
# 10. DATASET VALIDATION SUMMARY
# ------------------------------------------------------------

print("\nDATASET SUMMARY")
print("-" * 45)

print("Cases:", len(master))

print(
    "Distinct subjects:",
    master["subject_id"].nunique()
)

print(
    "Total annotated seizures:",
    master["seizures_verified"].sum()
)

print(
    "Unique recording files:",
    len(recordings_df)
)

print(
    "Total estimated recording hours:",
    round(master["recorded_hours"].sum(), 2)
)

print(
    "Recordings missing duration:",
    master["recordings_missing_duration"].sum()
)

print(
    "Cases missing age:",
    master["age_years"].isna().sum()
)

print(
    "Cases missing sex:",
    master["sex"].isna().sum()
)

# ------------------------------------------------------------
# 11. SAVE TABLES
# ------------------------------------------------------------

master.to_csv(
    "chbmit_patient_master.csv",
    index=False
)

all_seizures.to_csv(
    "chbmit_seizure_events.csv",
    index=False
)

recordings_df.to_csv(
    "chbmit_recording_metadata.csv",
    index=False
)

print("\nCSV files saved successfully.")
```
